# Importación de librerías
Para este caso vamos a utilizar dos de ellas:
- Numpy: para trabajar con matrices y vectores y operaciones con ellos.
- Matplotlib: para realizar la parte de graficación.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Definición de los datos de entrada
Aqui tenemos dos variables:}
- X es la variable independiente, para este caso el tamaño de una casa en metros cuadrados.
- y es la variable dependiente y en este caso es el precio de una casa en miles de pesos.

```python
X = np.array([50, 60, 70, 80, 90, 100, 110, 120])
y = np.array([150, 180, 195, 230, 260, 290, 325, 355])
```

```python
m = X.size
X_b = np.c_[np.ones((m, 1)), X]  # columna de unos (theta_0)
```

## Calculo de parámetros
Se localizan los parémetros óptimos que minimizan el error cuadratico medio:
- Theta[1]; significa la pendiente
- Theta[2]: se relaciona con el intercepto o sesgo.

```python
# Ecuacion normal: theta = (X^T X)^-1 X^T y
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y
print("Pendiente :", round(theta[1], 2))
print("Intercepto:", round(theta[0], 2))
```

```python
# Prediccion para una casa de 95 m^2
precio = theta[0] + theta[1] * 95
print("Precio estimado:", round(precio, 2), "miles")
```

## Graficación

Aquí vamos a generar el Scatterplot (gráfica de dispersión) y la linea e tendenciasegun la regresió linal calculada, esta última está de color azul.

```python
# Se genera la gráfica (Scatter) con Matplotlib:
plt.figure(figsize=(7,5))
plt.scatter(X,y, color = 'red',label = "Datos reales")
plt.show()
```

```python
x_l = np.linspace(X.min(), X.max(), 50)
y_l = theta[0] + theta[1] * x_l
```

```python
plt.figure(figsize=(7,5))
plt.scatter(X,y, color = 'red', label='Datos reales', s=60)
plt.plot(x_l, y_l, color = "blue", label = 'Regresión Lineal')
plt.title('Regresión lineal con el precio de un casa')
plt.ylabel('Precio (en miles)')
plt.xlabel('Tamaño ($m^2$)')
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```
